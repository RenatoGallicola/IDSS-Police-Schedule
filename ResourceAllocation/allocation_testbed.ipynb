```python
from resource_allocation import allocate_policemen as ap
import csv
import random
from location_generator import Location
```

```python
def __random_location_requirements(num_locations):
    return [random.randint(1, 5) for _ in range(num_locations)]
```

```python
# Max tolerated delay in minutes
tolerance = 10
```

```python
loc_data = []
with open('location_data.csv', newline='') as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        loc_data.append((float(row[0]), float(row[1])))

pol_data = []
with open('policeman_data.csv', newline='') as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        pol_data.append((str(row[0]), float(row[1]), float(row[2])))

loc_dist = [[round(((loc_data[j][0] - pol_data[i][1]) ** 2 + (loc_data[j][1] - pol_data[i][2]) ** 2) ** 0.5, Location.PRECISION.value) \
             for j in range(len(loc_data))] for i in range(len(pol_data))]

loc_req = __random_location_requirements(len(loc_data))

result = ap(loc_dist, loc_req, tolerance)
```

Find destination and travelled distance for each policeman

```python
if result:
    dest, dist = result
    print("First 10 policemen:")
    for i in range(10):
        print(f" + {pol_data[i][0]} -> Location: {dest[i]} - Distance: {dist[i]}")
else:
    print("No solution found.")
```

Check if the minimum required number of policeman is assigned to a location

```python
missing = []

for i in range(len(loc_data)):
    tot_police = 0
    for j in range(len(dest)):
       tot_police += 1 if dest[j] == i else 0
    if tot_police < loc_req[i]:
        missing.append(i)

if missing == []:
    print("All locations are covered")
else:
    print("The following locations are not covered: " + ', '.join(map(str, missing)))
```

Check which policeman will be late 

```python
# Average velocity of policemen in km/h
avg_vel = 30

# Time taken to reach location in minutes
time = [round(dist[i] / avg_vel * 60.0, 1) for i in range(len(dist))]

tot_late = 0

for i in range(len(time)):
    if time[i] > tolerance:
        print(f"{pol_data[i][0]} will take {time[i]} minutes to reach location {dest[i]}")
        tot_late += 1

print(f"Total late policemen: {tot_late}\n")
```
